## Compact results

Everything needed to rebuild the tables and charts locally, printed between two markers.

```python
import json, csv
rows = json.load(open("results/backtest.json"))
def r4(v): return round(v, 4) if isinstance(v, float) else v
compact = [{k: r4(v) for k, v in r.items()} for r in rows]
stations = list(csv.DictReader(open("results/stations.csv")))
print("<<<AURA_RESULTS>>>")
print(json.dumps({"backtest": compact, "stations": stations}, separators=(",", ":")))
print("<<<END_AURA_RESULTS>>>")
```
